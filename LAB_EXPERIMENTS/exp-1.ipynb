```python
# ==============================================================
# CSA4712 – DEEP LEARNING
# MS2 – SEPTEMBER 2026
#
# QUESTION 1:
# VECTOR AND MATRIX OPERATIONS FOR DEEP LEARNING
#
# LIBRARIES:
# NumPy
# Matplotlib
# ==============================================================


# ==============================================================
# 1. IMPORT LIBRARIES
# ==============================================================

import numpy as np
import matplotlib.pyplot as plt


# ==============================================================
# 2. CREATE DATASET
# ==============================================================

# Two sample feature vectors
# Features represent example deep-learning input values

feature_names = np.array([
    "Feature 1",
    "Feature 2",
    "Feature 3",
    "Feature 4",
    "Feature 5"
])

vector_A = np.array([10, 20, 30, 40, 50], dtype=float)
vector_B = np.array([5, 15, 25, 35, 45], dtype=float)


# ==============================================================
# 3. DISPLAY DATASET SAMPLE IN TABLE FORMAT
# ==============================================================

print("\n")
print("=" * 75)
print("DATASET SAMPLE")
print("=" * 75)

print(f"{'Index':<10}{'Feature':<20}{'Vector A':<20}{'Vector B':<20}")
print("-" * 75)

for i in range(len(feature_names)):
    print(
        f"{i + 1:<10}"
        f"{feature_names[i]:<20}"
        f"{vector_A[i]:<20.2f}"
        f"{vector_B[i]:<20.2f}"
    )

print("-" * 75)


# ==============================================================
# 4. BASIC DATASET STATISTICS
# ==============================================================

print("\n")
print("=" * 75)
print("DATASET STATISTICS")
print("=" * 75)

print(f"{'Statistic':<25}{'Vector A':<20}{'Vector B':<20}")
print("-" * 75)

print(
    f"{'Mean':<25}"
    f"{np.mean(vector_A):<20.2f}"
    f"{np.mean(vector_B):<20.2f}"
)

print(
    f"{'Median':<25}"
    f"{np.median(vector_A):<20.2f}"
    f"{np.median(vector_B):<20.2f}"
)

print(
    f"{'Minimum':<25}"
    f"{np.min(vector_A):<20.2f}"
    f"{np.min(vector_B):<20.2f}"
)

print(
    f"{'Maximum':<25}"
    f"{np.max(vector_A):<20.2f}"
    f"{np.max(vector_B):<20.2f}"
)

print(
    f"{'Standard Deviation':<25}"
    f"{np.std(vector_A):<20.2f}"
    f"{np.std(vector_B):<20.2f}"
)

print(
    f"{'Variance':<25}"
    f"{np.var(vector_A):<20.2f}"
    f"{np.var(vector_B):<20.2f}"
)

print("-" * 75)


# ==============================================================
# 5. VECTOR ADDITION
# ==============================================================

vector_addition = vector_A + vector_B

print("\n")
print("=" * 75)
print("VECTOR ADDITION")
print("=" * 75)

print(f"{'Feature':<20}{'A':<15}{'B':<15}{'A + B':<15}")
print("-" * 65)

for i in range(len(vector_A)):
    print(
        f"{feature_names[i]:<20}"
        f"{vector_A[i]:<15.2f}"
        f"{vector_B[i]:<15.2f}"
        f"{vector_addition[i]:<15.2f}"
    )

print("-" * 65)


# ==============================================================
# 6. VECTOR SUBTRACTION
# ==============================================================

vector_subtraction = vector_A - vector_B

print("\n")
print("=" * 75)
print("VECTOR SUBTRACTION")
print("=" * 75)

print(f"{'Feature':<20}{'A':<15}{'B':<15}{'A - B':<15}")
print("-" * 65)

for i in range(len(vector_A)):
    print(
        f"{feature_names[i]:<20}"
        f"{vector_A[i]:<15.2f}"
        f"{vector_B[i]:<15.2f}"
        f"{vector_subtraction[i]:<15.2f}"
    )

print("-" * 65)


# ==============================================================
# 7. ELEMENT-WISE MULTIPLICATION
# ==============================================================

elementwise_product = vector_A * vector_B

print("\n")
print("=" * 75)
print("ELEMENT-WISE VECTOR MULTIPLICATION")
print("=" * 75)

print(f"{'Feature':<20}{'A':<15}{'B':<15}{'A * B':<15}")
print("-" * 65)

for i in range(len(vector_A)):
    print(
        f"{feature_names[i]:<20}"
        f"{vector_A[i]:<15.2f}"
        f"{vector_B[i]:<15.2f}"
        f"{elementwise_product[i]:<15.2f}"
    )

print("-" * 65)


# ==============================================================
# 8. DOT PRODUCT
# ==============================================================

dot_product = np.dot(vector_A, vector_B)

print("\n")
print("=" * 75)
print("DOT PRODUCT")
print("=" * 75)

print("Vector A:", vector_A)
print("Vector B:", vector_B)
print("\nDot Product =", dot_product)


# ==============================================================
# 9. VECTOR NORMS
# ==============================================================

norm_A = np.linalg.norm(vector_A)
norm_B = np.linalg.norm(vector_B)

print("\n")
print("=" * 75)
print("VECTOR NORMS")
print("=" * 75)

print(f"{'Vector':<20}{'L2 Norm':<20}")
print("-" * 45)
print(f"{'Vector A':<20}{norm_A:<20.4f}")
print(f"{'Vector B':<20}{norm_B:<20.4f}")
print("-" * 45)


# ==============================================================
# 10. VECTOR NORMALIZATION
# ==============================================================

normalized_A = vector_A / norm_A
normalized_B = vector_B / norm_B

print("\n")
print("=" * 75)
print("VECTOR NORMALIZATION")
print("=" * 75)

print(f"{'Feature':<20}{'Normalized A':<20}{'Normalized B':<20}")
print("-" * 65)

for i in range(len(vector_A)):
    print(
        f"{feature_names[i]:<20}"
        f"{normalized_A[i]:<20.4f}"
        f"{normalized_B[i]:<20.4f}"
    )

print("-" * 65)


# ==============================================================
# 11. EUCLIDEAN DISTANCE
# ==============================================================

euclidean_distance = np.linalg.norm(vector_A - vector_B)

print("\n")
print("=" * 75)
print("EUCLIDEAN DISTANCE")
print("=" * 75)

print("Euclidean Distance between Vector A and Vector B:")
print(f"{euclidean_distance:.4f}")


# ==============================================================
# 12. COSINE SIMILARITY
# ==============================================================

cosine_similarity = np.dot(vector_A, vector_B) / (
    np.linalg.norm(vector_A) * np.linalg.norm(vector_B)
)

print("\n")
print("=" * 75)
print("COSINE SIMILARITY")
print("=" * 75)

print(f"Cosine Similarity = {cosine_similarity:.6f}")


# ==============================================================
# 13. SCALAR MULTIPLICATION
# ==============================================================

scalar = 2

scalar_result = scalar * vector_A

print("\n")
print("=" * 75)
print("SCALAR MULTIPLICATION")
print("=" * 75)

print(f"Scalar = {scalar}")

print(f"\n{'Feature':<20}{'Original A':<20}{'2 × A':<20}")
print("-" * 65)

for i in range(len(vector_A)):
    print(
        f"{feature_names[i]:<20}"
        f"{vector_A[i]:<20.2f}"
        f"{scalar_result[i]:<20.2f}"
    )

print("-" * 65)


# ==============================================================
# 14. CREATE MATRICES
# ==============================================================

matrix_A = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
], dtype=float)

matrix_B = np.array([
    [5, 15, 25],
    [35, 45, 55],
    [65, 75, 85]
], dtype=float)


# ==============================================================
# 15. DISPLAY MATRIX DATASET
# ==============================================================

print("\n")
print("=" * 75)
print("MATRIX DATASET")
print("=" * 75)

print("\nMatrix A:")
print(matrix_A)

print("\nMatrix B:")
print(matrix_B)


# ==============================================================
# 16. MATRIX ADDITION
# ==============================================================

matrix_addition = matrix_A + matrix_B

print("\n")
print("=" * 75)
print("MATRIX ADDITION")
print("=" * 75)

print(matrix_addition)


# ==============================================================
# 17. MATRIX SUBTRACTION
# ==============================================================

matrix_subtraction = matrix_A - matrix_B

print("\n")
print("=" * 75)
print("MATRIX SUBTRACTION")
print("=" * 75)

print(matrix_subtraction)


# ==============================================================
# 18. ELEMENT-WISE MATRIX MULTIPLICATION
# ==============================================================

matrix_elementwise = matrix_A * matrix_B

print("\n")
print("=" * 75)
print("ELEMENT-WISE MATRIX MULTIPLICATION")
print("=" * 75)

print(matrix_elementwise)


# ==============================================================
# 19. MATRIX MULTIPLICATION
# ==============================================================

matrix_multiplication = np.matmul(matrix_A, matrix_B)

print("\n")
print("=" * 75)
print("MATRIX MULTIPLICATION")
print("=" * 75)

print(matrix_multiplication)


# ==============================================================
# 20. MATRIX TRANSPOSE
# ==============================================================

matrix_transpose = matrix_A.T

print("\n")
print("=" * 75)
print("MATRIX TRANSPOSE")
print("=" * 75)

print(matrix_transpose)


# ==============================================================
# 21. MATRIX TRACE
# ==============================================================

matrix_trace = np.trace(matrix_A)

print("\n")
print("=" * 75)
print("MATRIX TRACE")
print("=" * 75)

print("Trace of Matrix A =", matrix_trace)


# ==============================================================
# 22. MATRIX RANK
# ==============================================================

matrix_rank = np.linalg.matrix_rank(matrix_A)

print("\n")
print("=" * 75)
print("MATRIX RANK")
print("=" * 75)

print("Rank of Matrix A =", matrix_rank)


# ==============================================================
# 23. INVERTIBLE MATRIX FOR DETERMINANT AND INVERSE
# ==============================================================

matrix_C = np.array([
    [4, 2, 1],
    [2, 5, 2],
    [1, 2, 4]
], dtype=float)


# ==============================================================
# 24. DETERMINANT
# ==============================================================

determinant_C = np.linalg.det(matrix_C)

print("\n")
print("=" * 75)
print("MATRIX DETERMINANT")
print("=" * 75)

print("Matrix C:")
print(matrix_C)

print(f"\nDeterminant of Matrix C = {determinant_C:.4f}")


# ==============================================================
# 25. MATRIX INVERSE
# ==============================================================

if not np.isclose(determinant_C, 0):

    inverse_C = np.linalg.inv(matrix_C)

    print("\n")
    print("=" * 75)
    print("MATRIX INVERSE")
    print("=" * 75)

    print("Inverse of Matrix C:")
    print(inverse_C)

else:

    print("Matrix C is singular and cannot be inverted.")


# ==============================================================
# 26. MATRIX STATISTICS
# ==============================================================

print("\n")
print("=" * 75)
print("MATRIX STATISTICS")
print("=" * 75)

print(f"{'Statistic':<25}{'Matrix A':<20}{'Matrix B':<20}")
print("-" * 70)

print(
    f"{'Mean':<25}"
    f"{np.mean(matrix_A):<20.2f}"
    f"{np.mean(matrix_B):<20.2f}"
)

print(
    f"{'Median':<25}"
    f"{np.median(matrix_A):<20.2f}"
    f"{np.median(matrix_B):<20.2f}"
)

print(
    f"{'Minimum':<25}"
    f"{np.min(matrix_A):<20.2f}"
    f"{np.min(matrix_B):<20.2f}"
)

print(
    f"{'Maximum':<25}"
    f"{np.max(matrix_A):<20.2f}"
    f"{np.max(matrix_B):<20.2f}"
)

print(
    f"{'Standard Deviation':<25}"
    f"{np.std(matrix_A):<20.2f}"
    f"{np.std(matrix_B):<20.2f}"
)

print(
    f"{'Variance':<25}"
    f"{np.var(matrix_A):<20.2f}"
    f"{np.var(matrix_B):<20.2f}"
)

print("-" * 70)


# ==============================================================
# 27. MATRIX-VECTOR MULTIPLICATION
# ==============================================================

input_vector = np.array([1, 2, 3], dtype=float)

matrix_vector_result = np.dot(matrix_A, input_vector)

print("\n")
print("=" * 75)
print("MATRIX-VECTOR MULTIPLICATION")
print("=" * 75)

print("Matrix A:")
print(matrix_A)

print("\nInput Vector:")
print(input_vector)

print("\nResult:")
print(matrix_vector_result)


# ==============================================================
# 28. VECTOR COMPARISON TABLE
# ==============================================================

print("\n")
print("=" * 75)
print("VECTOR COMPARISON AND EVALUATION")
print("=" * 75)

print(
    f"{'Metric':<30}"
    f"{'Vector A':<20}"
    f"{'Vector B':<20}"
    f"{'Comparison':<25}"
)

print("-" * 95)

mean_A = np.mean(vector_A)
mean_B = np.mean(vector_B)

std_A = np.std(vector_A)
std_B = np.std(vector_B)

max_A = np.max(vector_A)
max_B = np.max(vector_B)

min_A = np.min(vector_A)
min_B = np.min(vector_B)

print(
    f"{'Mean':<30}"
    f"{mean_A:<20.2f}"
    f"{mean_B:<20.2f}"
    f"{'A is higher':<25}"
)

print(
    f"{'Standard Deviation':<30}"
    f"{std_A:<20.2f}"
    f"{std_B:<20.2f}"
    f"{'A is higher':<25}"
)

print(
    f"{'Maximum':<30}"
    f"{max_A:<20.2f}"
    f"{max_B:<20.2f}"
    f"{'A is higher':<25}"
)

print(
    f"{'Minimum':<30}"
    f"{min_A:<20.2f}"
    f"{min_B:<20.2f}"
    f"{'A is higher':<25}"
)

print(
    f"{'L2 Norm':<30}"
    f"{norm_A:<20.4f}"
    f"{norm_B:<20.4f}"
    f"{'A has larger norm':<25}"
)

print("-" * 95)


# ==============================================================
# 29. DETAILED OPERATION EVALUATION TABLE
# ==============================================================

print("\n")
print("=" * 75)
print("DETAILED OPERATION RESULTS")
print("=" * 75)

print(f"{'Operation':<35}{'Result':<30}")
print("-" * 70)

print(
    f"{'Vector Addition Sum':<35}"
    f"{np.sum(vector_addition):<30.2f}"
)

print(
    f"{'Vector Subtraction Sum':<35}"
    f"{np.sum(vector_subtraction):<30.2f}"
)

print(
    f"{'Element-wise Product Sum':<35}"
    f"{np.sum(elementwise_product):<30.2f}"
)

print(
    f"{'Dot Product':<35}"
    f"{dot_product:<30.2f}"
)

print(
    f"{'Euclidean Distance':<35}"
    f"{euclidean_distance:<30.4f}"
)

print(
    f"{'Cosine Similarity':<35}"
    f"{cosine_similarity:<30.6f}"
)

print(
    f"{'Matrix A Trace':<35}"
    f"{matrix_trace:<30.2f}"
)

print(
    f"{'Matrix A Rank':<35}"
    f"{matrix_rank:<30}"
)

print(
    f"{'Matrix C Determinant':<35}"
    f"{determinant_C:<30.4f}"
)

print("-" * 70)


# ==============================================================
# 30. PLOT 1 – VECTOR COMPARISON
# ==============================================================

plt.figure(figsize=(10, 6))

x = np.arange(len(vector_A))

plt.plot(
    x,
    vector_A,
    marker='o',
    linewidth=2,
    label='Vector A'
)

plt.plot(
    x,
    vector_B,
    marker='s',
    linewidth=2,
    label='Vector B'
)

plt.title("Comparison of Vector A and Vector B")
plt.xlabel("Feature Index")
plt.ylabel("Feature Value")
plt.xticks(x, ["F1", "F2", "F3", "F4", "F5"])
plt.grid(True)
plt.legend()

plt.tight_layout()
plt.show()


# ==============================================================
# 31. PLOT 2 – VECTOR ADDITION AND SUBTRACTION
# ==============================================================

plt.figure(figsize=(10, 6))

plt.plot(
    x,
    vector_addition,
    marker='o',
    linewidth=2,
    label='A + B'
)

plt.plot(
    x,
    vector_subtraction,
    marker='s',
    linewidth=2,
    label='A - B'
)

plt.title("Vector Addition vs Vector Subtraction")
plt.xlabel("Feature Index")
plt.ylabel("Result")
plt.xticks(x, ["F1", "F2", "F3", "F4", "F5"])
plt.grid(True)
plt.legend()

plt.tight_layout()
plt.show()


# ==============================================================
# 32. PLOT 3 – VECTOR STATISTICAL COMPARISON
# ==============================================================

statistics_names = [
    "Mean",
    "Median",
    "Minimum",
    "Maximum",
    "Std Dev"
]

statistics_A = [
    np.mean(vector_A),
    np.median(vector_A),
    np.min(vector_A),
    np.max(vector_A),
    np.std(vector_A)
]

statistics_B = [
    np.mean(vector_B),
    np.median(vector_B),
    np.min(vector_B),
    np.max(vector_B),
    np.std(vector_B)
]

x_stats = np.arange(len(statistics_names))
width = 0.35

plt.figure(figsize=(11, 6))

plt.bar(
    x_stats - width / 2,
    statistics_A,
    width,
    label='Vector A'
)

plt.bar(
    x_stats + width / 2,
    statistics_B,
    width,
    label='Vector B'
)

plt.title("Statistical Comparison of Vectors")
plt.xlabel("Statistical Measure")
plt.ylabel("Value")
plt.xticks(x_stats, statistics_names)
plt.grid(axis='y')
plt.legend()

plt.tight_layout()
plt.show()


# ==============================================================
# 33. PLOT 4 – MATRIX A HEATMAP
# ==============================================================

plt.figure(figsize=(7, 6))

plt.imshow(
    matrix_A,
    interpolation='nearest'
)

plt.colorbar(label="Matrix Value")

plt.title("Matrix A Visualization")
plt.xlabel("Column")
plt.ylabel("Row")

plt.xticks([0, 1, 2], ["C1", "C2", "C3"])
plt.yticks([0, 1, 2], ["R1", "R2", "R3"])

plt.tight_layout()
plt.show()


# ==============================================================
# 34. PLOT 5 – MATRIX B HEATMAP
# ==============================================================

plt.figure(figsize=(7, 6))

plt.imshow(
    matrix_B,
    interpolation='nearest'
)

plt.colorbar(label="Matrix Value")

plt.title("Matrix B Visualization")
plt.xlabel("Column")
plt.ylabel("Row")

plt.xticks([0, 1, 2], ["C1", "C2", "C3"])
plt.yticks([0, 1, 2], ["R1", "R2", "R3"])

plt.tight_layout()
plt.show()


# ==============================================================
# 35. PLOT 6 – MATRIX MULTIPLICATION RESULT
# ==============================================================

plt.figure(figsize=(7, 6))

plt.imshow(
    matrix_multiplication,
    interpolation='nearest'
)

plt.colorbar(label="Value")

plt.title("Matrix Multiplication Result: A × B")
plt.xlabel("Column")
plt.ylabel("Row")

plt.xticks([0, 1, 2], ["C1", "C2", "C3"])
plt.yticks([0, 1, 2], ["R1", "R2", "R3"])

plt.tight_layout()
plt.show()


# ==============================================================
# 36. PLOT 7 – MATRIX STATISTICAL COMPARISON
# ==============================================================

matrix_statistics_names = [
    "Mean",
    "Median",
    "Minimum",
    "Maximum",
    "Std Dev"
]

matrix_statistics_A = [
    np.mean(matrix_A),
    np.median(matrix_A),
    np.min(matrix_A),
    np.max(matrix_A),
    np.std(matrix_A)
]

matrix_statistics_B = [
    np.mean(matrix_B),
    np.median(matrix_B),
    np.min(matrix_B),
    np.max(matrix_B),
    np.std(matrix_B)
]

x_matrix_stats = np.arange(len(matrix_statistics_names))

plt.figure(figsize=(11, 6))

plt.bar(
    x_matrix_stats - width / 2,
    matrix_statistics_A,
    width,
    label='Matrix A'
)

plt.bar(
    x_matrix_stats + width / 2,
    matrix_statistics_B,
    width,
    label='Matrix B'
)

plt.title("Statistical Comparison of Matrices")
plt.xlabel("Statistical Measure")
plt.ylabel("Value")
plt.xticks(
    x_matrix_stats,
    matrix_statistics_names
)
plt.grid(axis='y')
plt.legend()

plt.tight_layout()
plt.show()


# ==============================================================
# 37. EVALUATION
# ==============================================================

print("\n")
print("=" * 75)
print("EVALUATION OF RESULTS")
print("=" * 75)

print("\n1. VECTOR EVALUATION")
print("-" * 75)

print(
    f"Vector A has a mean value of {mean_A:.2f}, "
    f"while Vector B has a mean value of {mean_B:.2f}."
)

print(
    f"Vector A has a standard deviation of {std_A:.2f}, "
    f"while Vector B has a standard deviation of {std_B:.2f}."
)

print(
    f"The L2 norm of Vector A is {norm_A:.4f}, "
    f"whereas Vector B has an L2 norm of {norm_B:.4f}."
)

print(
    f"The Euclidean distance between the two vectors is "
    f"{euclidean_distance:.4f}."
)

print(
    f"The cosine similarity between the two vectors is "
    f"{cosine_similarity:.6f}."
)


print("\n2. MATRIX EVALUATION")
print("-" * 75)

print(
    f"Matrix A has a mean value of {np.mean(matrix_A):.2f} "
    f"and Matrix B has a mean value of {np.mean(matrix_B):.2f}."
)

print(
    f"The rank of Matrix A is {matrix_rank}."
)

print(
    f"The trace of Matrix A is {matrix_trace:.2f}."
)

print(
    f"The determinant of Matrix C is {determinant_C:.4f}."
)

print(
    "Since the determinant of Matrix C is non-zero, "
    "Matrix C is invertible."
)


# ==============================================================
# 38. FINAL COMPARISON
# ==============================================================

print("\n")
print("=" * 75)
print("FINAL COMPARISON")
print("=" * 75)

print(
    f"{'Parameter':<30}"
    f"{'Vector/Matrix A':<20}"
    f"{'Vector/Matrix B':<20}"
)

print("-" * 75)

print(
    f"{'Mean':<30}"
    f"{np.mean(vector_A):<20.2f}"
    f"{np.mean(vector_B):<20.2f}"
)

print(
    f"{'Standard Deviation':<30}"
    f"{np.std(vector_A):<20.2f}"
    f"{np.std(vector_B):<20.2f}"
)

print(
    f"{'Maximum':<30}"
    f"{np.max(vector_A):<20.2f}"
    f"{np.max(vector_B):<20.2f}"
)

print(
    f"{'Minimum':<30}"
    f"{np.min(vector_A):<20.2f}"
    f"{np.min(vector_B):<20.2f}"
)

print("-" * 75)


# ==============================================================
# 39. FINAL CONCLUSION
# ==============================================================

print("\n")
print("=" * 75)
print("CONCLUSION")
print("=" * 75)

print(
    "Vector and matrix operations were successfully implemented "
    "using NumPy."
)

print(
    "The experiment demonstrated vector addition, subtraction, "
    "multiplication, dot product, normalization, norms and "
    "similarity calculations."
)

print(
    "Matrix addition, subtraction, element-wise multiplication, "
    "matrix multiplication, transpose, rank, trace, determinant "
    "and inverse were also performed."
)

print(
    "Statistical analysis and graphical comparisons were used "
    "to evaluate the generated results."
)

print(
    "The visualizations provide a clear comparison of vector "
    "and matrix values and their corresponding operations."
)

print("\n")
print("=" * 75)
print("PROGRAM EXECUTED SUCCESSFULLY")
print("=" * 75)
```
