```python

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm, skew

np.random.seed(42)

# Generate skewed population
population = np.random.exponential(scale=10, size=100000)

sizes = [5, 30, 100]
samples = 1000

fig, axes = plt.subplots(1, 3, figsize=(13, 4))

for ax, n in zip(axes, sizes):
    means = [np.mean(np.random.choice(population, n))
             for _ in range(samples)]

    # Sampling distribution
    ax.hist(means, bins=30, density=True, alpha=0.6)

    # Gaussian curve
    x = np.linspace(min(means), max(means), 100)
    ax.plot(x, norm.pdf(x, np.mean(means),
                        np.std(means)), 'r')

    ax.set_title(f'Sample Size = {n}')
    ax.set_xlabel('Sample Mean')
    ax.set_ylabel('Density')

    print("Sample size:", n)
    print("Mean:", round(np.mean(means), 3))
    print("Standard deviation:", round(np.std(means), 3))
    print("Skewness:", round(skew(means), 3))

plt.tight_layout()
plt.show()
```
