```python

import numpy as np
from scipy.linalg import inv, det

# 1. Construct Matrices
A = np.array([[2, 1, 3],
              [1, 2, 2],
              [3, 1, 4]])

B = np.array([[1, 2],
              [2, 4],
              [3, 6]])

C = np.array([[2, 1, 0],
              [1, 3, 2],
              [0, 1, 2]])

# 2. Matrix Rank
print("Rank of A:", np.linalg.matrix_rank(A))
print("Rank of B:", np.linalg.matrix_rank(B))
print("Rank of C:", np.linalg.matrix_rank(C))

# 3. Determinant
print("\nDeterminant of A:", det(A))
print("Determinant of C:", det(C))

# 4. Check Invertibility
for name, M in [("A", A), ("C", C)]:
    d = np.linalg.det(M)
    print("\nMatrix", name)

    if not np.isclose(d, 0):
        print("Invertible")
        M_inv = inv(M)
        print("Inverse:\n", M_inv)
        print("Identity Product:\n", M @ M_inv)
    else:
        print("Not Invertible")

# 5. Linearly Dependent Columns
D = np.array([[1, 2, 3],
              [2, 4, 6],
              [3, 6, 9]])

print("\nDependent Matrix:\n", D)
print("Rank:", np.linalg.matrix_rank(D))
print("Determinant:", det(D))

if np.linalg.matrix_rank(D) == D.shape[0]:
    print("Invertible")
else:
    print("Not Invertible")
```
