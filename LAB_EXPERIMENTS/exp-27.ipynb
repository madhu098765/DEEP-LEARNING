```python

import numpy as np
import matplotlib.pyplot as plt

# 1. Customer-Genre Matrix (6x5)
C = np.array([
    [5, 4, 3, 2, 1],
    [4, 5, 2, 1, 3],
    [1, 2, 5, 4, 3],
    [2, 1, 4, 5, 3],
    [5, 3, 4, 2, 1],
    [1, 3, 2, 5, 4]
])

# 2. Genre-Book Matrix (5x4)
B = np.array([
    [0.8, 0.2, 0.5, 0.1],
    [0.3, 0.9, 0.2, 0.4],
    [0.5, 0.3, 0.8, 0.2],
    [0.2, 0.4, 0.3, 0.9],
    [0.1, 0.5, 0.4, 0.6]
])

print("Customer Matrix:\n", C)
print("\nBook Matrix:\n", B)

# 3. Dot Product
print("\nDot Product:", np.dot(C[0], B[:, 0]))

# 4. Matrix Multiplication
S = np.dot(C, B)
print("\nCustomer-Book Score Matrix:\n", S)

# 5. Transpose
print("\nTranspose:\n", C.T)

# 6. Inverse (for valid square matrix)
A = np.array([[2, 1], [1, 2]])
print("\nInverse:\n", np.linalg.inv(A))

# 7. Element-wise Operations
print("\nAddition:\n", C + C)
print("\nMultiplication:\n", C * C)

# 8. Cosine Similarity
norm = np.linalg.norm(C, axis=1)
similarity = C @ C.T / np.outer(norm, norm)
print("\nCosine Similarity:\n", np.round(similarity, 3))

np.fill_diagonal(similarity, -1)
i, j = np.unravel_index(np.argmax(similarity), similarity.shape)
print("\nMost Similar Customers:", i+1, "and", j+1)

# 9. Dense Layer: y = Wx + b
x = C[0]
W = np.array([[0.2, 0.3, 0.1, 0.4, 0.5],
              [0.1, 0.2, 0.3, 0.2, 0.1]])
bias = np.array([1, 2])

y = W @ x + bias
print("\nDense Layer Output:", y)

# 10. Heatmap
plt.imshow(S, cmap="viridis", aspect="auto")
plt.colorbar(label="Recommendation Score")
plt.xlabel("Books")
plt.ylabel("Customers")
plt.title("Customer-Book Recommendation Scores")
plt.xticks(range(4), ["Book 1", "Book 2", "Book 3", "Book 4"])
plt.yticks(range(6), ["C1", "C2", "C3", "C4", "C5", "C6"])
plt.show()
```
