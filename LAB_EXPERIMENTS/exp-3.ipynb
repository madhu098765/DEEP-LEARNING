```python
import numpy as np
import matplotlib.pyplot as plt

from scipy.stats import norm, binom, poisson

# 1. Normal Distribution
x = np.linspace(-4, 4, 500)

normal_values = norm.pdf(
    x,
    loc=0,
    scale=1
)

print("NORMAL DISTRIBUTION")

print("P(X <= 1):", norm.cdf(1))

print(
    "P(-1 <= X <= 1):",
    norm.cdf(1) - norm.cdf(-1)
)

# 2. Binomial Distribution
n = 10
p = 0.4

k = np.arange(0, n + 1)

binomial_values = binom.pmf(
    k,
    n,
    p
)

print("\nBINOMIAL DISTRIBUTION")

print(
    "Probability of exactly 4 successes:",
    binom.pmf(4, n, p)
)

# 3. Poisson Distribution
lam = 4

k_poisson = np.arange(0, 15)

poisson_values = poisson.pmf(
    k_poisson,
    lam
)

print("\nPOISSON DISTRIBUTION")

print(
    "Probability of 3 events:",
    poisson.pmf(3, lam)
)

# 4. Bayes Theorem
prevalence = 0.02
sensitivity = 0.95
false_positive = 0.05

# P(Positive)
p_positive = (
    sensitivity * prevalence
    + false_positive * (1 - prevalence)
)

# P(Condition | Positive)
posterior = (
    sensitivity * prevalence
    / p_positive
)

print("\nBAYES THEOREM")

print("Prior Probability:", prevalence)
print("Sensitivity:", sensitivity)
print("False Positive Rate:", false_positive)

print(
    "Posterior Probability:",
    posterior
)

# 5. Visualization
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.plot(x, normal_values)
plt.title("Normal Distribution")
plt.xlabel("X")
plt.ylabel("Probability Density")

plt.subplot(1, 3, 2)
plt.bar(k, binomial_values)
plt.title("Binomial Distribution")
plt.xlabel("Successes")

plt.subplot(1, 3, 3)
plt.bar(k_poisson, poisson_values)
plt.title("Poisson Distribution")
plt.xlabel("Events")

plt.tight_layout()
plt.show()
```
